```python
#Latihan 4

tegangan_harian = [217.5, 219.0, 220.3, 218.7, 221.1, 216.9]

# (a) Cetak pembacaan pertama dan terakhir menggunakan indeks
print("Pembacaan pertama:", tegangan_harian[0])
print("Pembacaan terakhir:", tegangan_harian[-1])

# (b) Tambahkan pembacaan baru
tegangan_harian.append(222.0)

# (c) Cetak jumlah total data
print("Jumlah total data:", len(tegangan_harian))
```

```python
#Latihan 5

# Membuat dictionary spesifikasi motor listrik
motor = {
    "nama": "Motor Induksi 3 Fasa",
    "daya_kW": 5.5,
    "tegangan_V": 380,
    "rpm": 1450

}

# Mencetak salah satu nilai menggunakan key
print("Daya motor:", motor["daya_kW"], "kW")

# Mengubah nilai rpm
motor["rpm"] = 1500

# Mencetak seluruh dictionary
print("Spesifikasi motor setelah diubah:")
print(motor)
```

```python
# Jawaban Latihan 6

pembangkit = [
    {"nama": "Unit-1", "kapasitas_MW": 100, "status": "operasi"},
    {"nama": "Unit-2", "kapasitas_MW": 150, "status": "operasi"},
    {"nama": "Unit-3", "kapasitas_MW": 80,  "status": "pemeliharaan"},
]

# Cetak nama unit kedua
print("Nama unit kedua:", pembangkit[1]["nama"])

# Ubah status salah satu unit
pembangkit[1]["status"] = "pemeliharaan"

# Tambahkan satu unit baru
pembangkit.append({
    "nama": "Unit-4",
    "kapasitas_MW": 120,
    "status": "operasi"
})

# Cetak seluruh data
print("Data seluruh pembangkit:")
print(pembangkit)

```
